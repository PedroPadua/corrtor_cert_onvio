```python
import pandas as pd
import re
def _sanatize_cnpj(cnpj):
    return re.sub(r"\D", "", str(cnpj))
```

```python
df = pd.read_excel(r"C:\Users\ia01\Downloads\Procuracoes Acessorias 1.xlsx", sheet_name = 'Vencimentos')
```

```python
df = df.rename(columns={'EmpID': 'id', 'Empresa': 'emp', 'CNPJ':'cnpj', 'OBRIGAÇÃO/TAREFA': 'tarefa', 'Prazo legal':'data'})
df.columns.to_list()
```

```python
len(df)
df['cnpj'] = df['cnpj'].apply(_sanatize_cnpj)
df.head()
```

```python
df = df.sort_values('cnpj')
```

```python
for nome, sub in df.groupby('emp'):
    print(nome)       
    print(len(sub['tarefa']))
```
